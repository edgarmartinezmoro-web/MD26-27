```python
def torre_hanoi(n, origen, destino, auxiliar):
    # Caso base: si solo hay un disco, muévelo directamente
    if n == 1:
        print(f"Mover disco 1 de {origen} a {destino}")
        return
    
    # 1. Mover n-1 discos de la torre origen a la auxiliar
    torre_hanoi(n - 1, origen, auxiliar, destino)
    
    # 2. Mover el disco restante (el más grande) de origen a destino
    print(f"Mover disco {n} de {origen} a {destino}")
    
    # 3. Mover los n-1 discos de la auxiliar a la torre destino
    torre_hanoi(n - 1, auxiliar, destino, origen)

# Ejemplo con 3 discos
discos = 12
torre_hanoi(discos, "Origen", "Destino", "Auxiliar")
```
