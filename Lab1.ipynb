e) $
p \rightarrow (q \rightarrow r)
$

```python
def impl(p, q):
    ### Implicación p -> q (equivalente a no p o q)
    return int(not p or q)

def formula(p, q, r):
    ### p -> (q -> r)
    return impl(p, impl(q, r))

print(f"{'p':<6}{'q':<6}{'r':<6}{'p -> (q -> r)':<15}")
print("-" * 33)

for p in [0, 1]:
    for q in [0, 1]:
        for r in [0, 1]:
            resultado = formula(p, q, r)
            print(f"{p:<6}{q:<6}{r:<6}{resultado:<15}")
```

g) $\neg(p \vee \neg q) \rightarrow \neg p$

```python
def formula3(p, q):
    ### ~(p v ~q) -> ~p
    antecedente = int(not (p or (not q)))
    consecuente = int(not p)
    return impl(antecedente, consecuente)

print(f"{'p':<6}{'q':<6}{'~(p v ~q) -> ~p':<20}")
print("-" * 32)

for p in [0, 1]:
    for q in [0, 1]:
        resultado = formula3(p, q)
        print(f"{p:<6}{q:<6}{resultado:<20}")
```

1.9.a)i) Simplifica $\neg [(p \vee q) \rightarrow r]$

```python
from sympy import symbols, Not, And, Or, Implies, latex, simplify_logic
from sympy.logic.boolalg import to_nnf

# Definición de variables
p, q, r = symbols('p q r')

# Expresión: ~((p | q) -> r)
expr1 = Not(Implies(Or(p, q), r))

def simplificar_paso_a_paso(expresion):
    print("--- PASO 0: Expresión Original ---")
    print(f"Python: {expresion}")
    
    # Paso 1: Eliminar la implicación interna (A -> B  ≡  ~A | B)
    paso1 = expresion.replace(Implies, lambda a, b: Or(Not(a), b))
    print("--- PASO 1: Eliminación de la Implicación ---")
    print(f"Python: {paso1}")
    
    # Paso 2: Introducir la negación mediante Leyes de De Morgan (Forma Normal Negativa)
    paso2 = to_nnf(paso1)
    print("--- PASO 2: Aplicación de Leyes de De Morgan ---")
    print(f"Python: {paso2}")

    # Paso 3: Simplificación de la conjunción
    paso3 = simplify_logic(paso2)
    print("--- PASO 3: Simplificación Final ---")
    print(f"Python: {paso3}")
    
# Ejecutar el procedimiento
simplificar_paso_a_paso(expr1)
```

1.9.a)ii) Simplifica $(p \vee q) \wedge \neg (\neg p \wedge q)$

```python
#variables
p, q = symbols('p q')
# Expresión: (p | q) & ~(~p & q)
expr2 = And(Or(p, q), Not(And(Not(p), q)))
# Ejecutar el procedimiento
simplificar_paso_a_paso(expr2)
```

1.9.a)iii) Simplifica $\neg(\neg[(p \vee q) \wedge r] \vee \neg q)$

```python
p, q, r = symbols('p q r')
# Expresión: ~ (~[(p | q) & r] | ~q)
expr3 = Not(Or(Not(And(Or(p, q), r)), Not(q)))
simplificar_paso_a_paso(expr3)
```

1.11.a) $\left[ (p \to q) \land (r \to s) \right] \to \left[ (p \lor r) \to (q \lor s) \right]$

```python
from sympy import And, Implies, Or, simplify_logic, symbols

# Definición de las variables proposicionales
p, q, r, s = symbols("p q r s")

# Expresión original: [(p -> q) ^ (r -> s)] -> [(p v r) -> (q v s)]
expr4 = Implies(And(Implies(p, q), Implies(r, s)), Implies(Or(p, r), Or(q, s)))

# Simplificación de la expresión
simp = simplify_logic(expr4)
print(f"Python: {simp}")
```

1.11.b) $\left[ (p \to q) \land \neg q \right] \to p$

```python
from sympy import And, Implies, Not, simplify_logic, symbols

# Definición de las variables proposicionales
p, q = symbols("p q")

# Expresión original: [(p -> q) ^ ~q] -> p
expr5 = Implies(And(Implies(p, q), Not(q)), p)

# Simplificación de la expresión
simp = simplify_logic(expr5)
print(f"Python: {simp}")
```
