<b>Ejemplo (de juguete): Problema de la isla y los ojos azules.</b>

En una isla vive un grupo de personas de ojos de distintos colores. Todos son lógicos perfectos: si una conclusión se puede deducir, la descubren al instante.

Nadie sabe el color de sus propios ojos y está prohibido comunicarse. Sin embargo, todos se ven entre sí y llevan la cuenta de los colores de los demás. Cada medianoche, un barco llega a la isla: cualquiera que haya deducido el color de sus propios ojos debe subir a él y marcharse; los demás se quedan. Todos conocen estas reglas.

En la isla hay 100 personas de ojos azules, 100 de ojos marrones y la Gurú (que tiene ojos verdes). Por tanto, cualquier persona de ojos azules ve a 100 personas de ojos marrones y a 99 de ojos azules, pero eso no le basta para saber su propio color.

Un día a mediodía, la Gurú se sitúa ante todos y hace una única declaración pública: "Veo a alguien que tiene los ojos azules".

**Pregunta: ¿Quiénes se van de la isla y en qué noche lo hacen?**

(Nota: No hay trucos, trampas ni reflejos. La respuesta es puramente lógica). 
</div>

**Simulación con Python**

```python
def resolver_acertijo_isla(num_azules, num_marrones):
    print(f"--- SIMULACIÓN DEL ACERTIJO DE LA ISLA ---")
    print(f"Población: {num_azules} ojos azules, {num_marrones} ojos marrones, 1 Guru (ojos verdes).\n")
    
    # Cada persona con ojos azules ve a (num_azules - 1) personas con ojos azules.
    # Cada persona con ojos marrones ve a num_azules personas con ojos azules.
    azules_vistos_por_azul = num_azules - 1
    azules_vistos_por_marron = num_azules
    
    # El ferry llega cada noche. Evaluamos día a día.
    for dia in range(1, num_azules + 2):
        # Regla lógica: Si nadie se fue en la noche 'dia - 1', se vuelve conocimiento
        # público que el número total de personas con ojos azules es mayor que 'dia - 1'.
        
        # Un isleño deduce su color si el número de ojos azules que ve es igual a 'dia - 1'
        azul_deduce = (azules_vistos_por_azul == dia - 1)
        marron_deduce = (azules_vistos_por_marron == dia - 1)
        
        if azul_deduce:
            print(f"-> ¡NOCHE {dia}! Las {num_azules} personas con ojos azules descubren su color y DEJAN la isla.")
            print(f"   Razón: Cada uno ve exactamente {azules_vistos_por_azul} personas con ojos azules.")
            print(f"   Como nadie se fue en la noche {dia-1}, deducen que el total es al menos {dia}.")
            print(f"   Por lo tanto, ellos mismos completan el grupo.\n")
            break
        else:
            if dia <= num_azules:
                print(f"Noche {dia}: Nadie se marcha. Ahora todos saben que hay al menos {dia + 1} personas con ojos azules.")

    # Estado de las personas con ojos marrones tras la partida
    dia_siguiente = num_azules + 1
    print(f"--- DÍA {dia_siguiente} (Post-partida) ---")
    print(f"Las personas con ojos marrones ven que los de ojos azules se marcharon en la noche {num_azules}.")
    print(f"Deducción de los de ojos marrones: 'Sé con certeza que NO tengo ojos azules'.")
    print(f"Resultado: Como ven {num_marrones - 1} personas con ojos marrones y 1 con ojos verdes,")
    print(f"           siguen sin saber si sus propios ojos son marrones o de otro color desconocido (ej. rojos).")
    print(f"           Por lo tanto, las {num_marrones} personas de ojos marrones y la Guru SE QUEDAN en la isla.")

# Ejecutamos la simulación para los 100 de ojos azules y 100 de ojos marrones
resolver_acertijo_isla(num_azules=100, num_marrones=100)
```

<b>Ejemplo (de juguete): Configuraciones de un cubo de Rubik</b>

¿Cuál es el número total de configuraciones esencialmente diferentes (no equivalentes por rotación espacial) de un cubo de Rubik 3x3?
</div>

```python
import math

def calcular_estados_centros_fijos():
    # 1. Elementos móviles del cubo (los 6 centros no se mueven)
    esquinas = 8
    aristas = 12
    
    # 2. Permutaciones (posiciones físicas que pueden ocupar)
    perm_esquinas = math.factorial(esquinas)   # 8!
    perm_aristas = math.factorial(aristas)     # 12!
    
    # 3. Orientaciones (giros sobre su propio eje de cada pieza)
    # Ley de paridad: la última pieza depende de las anteriores
    orient_esquinas = 3 ** (esquinas - 1)      # 3^7
    orient_aristas = 2 ** (aristas - 1)        # 2^11
    
    # 4. Paridad global de permutación
    # Solo la mitad de los estados de posición son alcanzables mediante giros
    paridad_global = 2
    
    # 5. Cálculo final con centros fijos
    estados_totales = (perm_esquinas * perm_aristas * orient_esquinas * orient_aristas) // paridad_global
    
    return estados_totales

# Ejecución
estados = calcular_estados_centros_fijos()
print(f"Número de estados con centros fijos: {estados:,}")
```

<b>Ejemplo: Ruta más corta</b>

Encuentra la ruta más corta entre dos vértices de un grafo.
</div>

```python
import networkx as nx
import matplotlib.pyplot as plt

# 1. Crear el grafo y añadir las conexiones con sus pesos
G = nx.Graph()
conexiones = [
    ('A', 'B', 4),
    ('A', 'C', 2),
    ('B', 'C', 5),
    ('B', 'D', 10),
    ('C', 'D', 3)
]
G.add_weighted_edges_from(conexiones)

# 2. Configurar el lienzo del dibujo
plt.figure(figsize=(8, 6))

# Calcular posiciones fijas para que el diseño sea siempre idéntico y limpio
pos = nx.spring_layout(G, seed=42)

# 3. Dibujar los componentes del grafo de forma neutral
# Nodos (puntos)
nx.draw_networkx_nodes(G, pos, node_color='lightgray', node_size=900, edgecolors='black')

# Etiquetas de los nodos (letras A, B, C, D)
nx.draw_networkx_labels(G, pos, font_size=14, font_family='sans-serif', font_weight='bold')

# Aristas (líneas de conexión)
nx.draw_networkx_edges(G, pos, edge_color='black', width=2)

# Etiquetas de los pesos (valores numéricos en los caminos)
etiquetas_pesos = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(G, pos, edge_labels=etiquetas_pesos, font_size=12, font_weight='bold')

# 4. Ajustes finales de visualización
plt.title("Enunciado: Encuentra el camino más corto de A a D", fontsize=14, pad=20)
plt.axis('off')  # Elimina el recuadro y los ejes numéricos de fondo
plt.tight_layout()

# Mostrar el gráfico en pantalla
plt.show()
```

```python
# 2. Calcular la ruta más corta (Dijkstra)
origen, destino = 'A', 'D'
ruta = nx.shortest_path(G, source=origen, target=destino, weight='weight')
distancia = nx.shortest_path_length(G, source=origen, target=destino, weight='weight')

print(f"Ruta óptima: {' -> '.join(ruta)} (Distancia total: {distancia})")

# 3. Dibujar el grafo
plt.figure(figsize=(8, 6))
pos = nx.spring_layout(G, seed=42)  # Posición fija de los nodos

# Dibujar todos los nodos y aristas base
nx.draw_networkx_nodes(G, pos, node_color='lightblue', node_size=800)
nx.draw_networkx_labels(G, pos, font_size=14, font_weight='bold')
nx.draw_networkx_edges(G, pos, edgelist=G.edges(), edge_color='gray', width=2)

# Crear la lista de aristas que forman la ruta más corta
aristas_ruta = list(zip(ruta, ruta[1:]))

# Resaltar la ruta más corta en color rojo y más gruesa
nx.draw_networkx_edges(G, pos, edgelist=aristas_ruta, edge_color='red', width=4)

# Dibujar las etiquetas de los pesos en los caminos
etiquetas_pesos = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(G, pos, edge_labels=etiquetas_pesos, font_size=12)

plt.title(f"Ruta más corta de {origen} a {destino} (Resaltada en Rojo)", fontsize=14)
plt.axis('off')  # Ocultar los ejes numéricos del gráfico
plt.show()
```

<b>Ejemplo: Calendario perpetuo</b>

Por ejemplo, ¿Qué día de la semana fue el 20 de julio de 1969 (el día que el ser humano pisó la Luna)?
</div>

```python
def algoritmo_zeller(dia, mes, ano):
    """
    Calcula el día de la semana para cualquier fecha del calendario gregoriano.
    Entradas: dia (int), mes (int), ano (int)
    Devuelve: Cadena con el nombre del día en castellano.
    """
    # Ajuste para Enero y Febrero (se consideran meses 13 y 14 del año anterior)
    if mes < 3:
        mes += 12
        ano -= 1
        
    K = ano % 100    # Año dentro del siglo (ej. 69 para 1969)
    J = ano // 100   # El siglo (ej. 19 para 1969)
    
    # Fórmula oficial de la Congruencia de Zeller
    # h devuelve: 0=Sábado, 1=Domingo, 2=Lunes, 3=Martes, 4=Miércoles, 5=Jueves, 6=Viernes
    h = (dia + ((13 * (mes + 1)) // 5) + K + (K // 4) + (J // 4) - (2 * J)) % 7
    
    # Conversión del resultado 'h' a nuestro orden semanal estándar
    dias_semana = {
        1: "Domingo",
        2: "Lunes",
        3: "Martes",
        4: "Miércoles",
        5: "Jueves",
        6: "Viernes",
        0: "Sábado"
    }
    
    return dias_semana[h]
```

```python
# Probamos con la fecha histórica anterior: 20 de Julio de 1969
dia_test = 20
mes_test = 7
ano_test = 1969

resultado = algoritmo_zeller(dia_test, mes_test, ano_test)
print(f"El {dia_test}/{mes_test}/{ano_test} fue: {resultado}")
```

1.- **Recuerda de bachiller:** ¿Qué dos formas hay de resolver sistemas de ecuaciones lineales para sistemas compatibles determinados?

```python
import matplotlib.pyplot as plt
import numpy as np
import math

# 1. Rango limitado hasta n=11 debido al crecimiento factorial
n = np.arange(2, 12)

# 2. Operaciones para Eliminación de Gauss O(n^3)
ops_gauss = (2/3) * (n**3) + n**2

# 3. Operaciones para Cramer con Laplace O((n+1)!)
ops_cramer_laplace = np.array([math.factorial(x + 1) for x in n])

# 4. Configuración y renderizado de la gráfica
plt.figure(figsize=(10, 6))

plt.plot(n, ops_gauss, label=r"Eliminación de Gauss $\mathcal{O}(n^3)$", color="blue", linewidth=2.5, marker='o')
plt.plot(n, ops_cramer_laplace, label=r"Cramer Puro (Laplace) $\mathcal{O}((n+1)!)$", color="darkred", linewidth=2.5, linestyle="--", marker='s')

# Es obligatorio usar escala logarítmica
plt.yscale("log")

plt.title("Cramer Puro vs Eliminación de Gauss", fontsize=14, fontweight="bold")
plt.xlabel(r"Número de Incógnitas ($n$)", fontsize=12)
plt.ylabel("Operaciones Estimadas (Escala Log)", fontsize=12)

plt.grid(True, which="both", linestyle=":", alpha=0.6)
plt.legend(fontsize=12, loc="upper left")

# Datos analíticos
print(f"Comparativa para n = 11:")
print(f" - Gauss requiere:                  {ops_gauss[-1]:,.0f} operaciones.")
print(f" - Cramer Puro (Laplace) requiere:  {ops_cramer_laplace[-1]:,.0f} operaciones.")

plt.show()
```

```python
n = 6309573444801942599499775
# Consideramos el número n^4+4
# Factorización de Sophie Germain
factor_A = n**2 - 2*n + 2
factor_B = n**2 + 2*n + 2

# El número N de 100 dígitos
N = factor_A * factor_B

# Asumimos que P es un factor primo extraído de la componente (Factor A)
P = factor_A  

# 2. Creamos los números 
numero_1 = N**2    
numero_2 = P**3

print(f"Número_1: {(str(numero_1))}")
print(f"Dígitos de número_1: {len(str(numero_1))}")
print(f"Número_2: {(str(numero_2))}")
print(f"Dígitos de número_2: {len(str(numero_2))}")
```

3.- **Evaluar un polinomio**

$$
P(x)=7x^{23}-4x^{22}+12x^{21}-9x^{20}+3x^{19}+15x^{18}-6x^{17}+8x^{16}-11x^{15}+5x^{14}-2x^{13}+14x^{12}\newline -7x^{11}+x^{10}-13x^9+6x^8-10x^7+4x^6-8x^5+9x^4-3x^3+11x^2-5x+2
$$



```python
# Coeficientes del polinomio
coef = [
    7, -4, 12, -9, 3, 15, -6, 8, -11, 5, -2, 14,
    -7, 1, -13, 6, -10, 4, -8, 9, -3, 11, -5, 2
]

def horner(coeficientes, x):
    resultado = coeficientes[0]
    
    for c in coeficientes[1:]:
        resultado = resultado * x + c
    
    return resultado

# Valor donde evaluar
x0 = 2

valor = horner(coef, x0)

print("P(", x0, ") =", valor)
```

Su **expresión de Horner-Ruffini** es:

$$
P(x)=(((((((((((((((((((((((7x-4)x+12)x-9)x+3)x+15)x-6)x+8)x-11)x+5)\ldots \newline \ldots x-2)x+14)x-7)x+1)x-13)x+6)x-10)x+4)x-8)x+9)x-3)x+11)x-5)x+2
$$

```python
import matplotlib.pyplot as plt

# Grados de los polinomios
grados = range(1, 51)

# Aproximación del número de operaciones
# Método usual: suma de potencias + productos
operaciones_normal = [n*(n+1)//2 + n for n in grados]

# Horner-Ruffini: una multiplicación y una suma por coeficiente
operaciones_horner = [2*n for n in grados]

# Crear gráfica
plt.figure(figsize=(10, 6))

plt.plot(grados, operaciones_normal, label="Evaluación usual")
plt.plot(grados, operaciones_horner, label="Horner-Ruffini")

plt.xlabel("Grado del polinomio (n)")
plt.ylabel("Número aproximado de operaciones")
plt.title("Comparación de complejidad: Evaluación usual vs Horner-Ruffini")

plt.legend()
plt.grid(True)

plt.show()
```
